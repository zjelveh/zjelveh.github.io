## Part 1: Boolean Values and Comparisons

A **Boolean** value answers a yes-or-no question:

- `True` means the condition is met.
- `False` means the condition is not met.

`True` and `False` are Python values. They begin with capital letters and do not use quotation marks. They do not mean that the data are good or bad. They report whether a specific condition is true.

A single equals sign, `=`, assigns a value to a variable. A comparison operator asks a question about values and produces `True` or `False`.

| Operator | Question |
|---|---|
| `==` | Are these values equal? |
| `!=` | Are these values not equal? |
| `>` | Is the first value greater than the second? |
| `<` | Is the first value less than the second? |
| `>=` | Is the first value greater than or equal to the second? |
| `<=` | Is the first value less than or equal to the second? |

````python
state = "Virginia"
state == "Virginia"
````

````python
state != "Virginia"
````

````python
property_rate = 1642.7
property_rate >= 1500
````

````python
# String comparisons are case-sensitive
"Virginia" == "virginia"
````

Each expression asks one question without changing the original values. Next class, we will compare an entire DataFrame column with a value. Pandas will return `True` or `False` for every row and use those results to select records.

## Part 2: Why Move Beyond Lists?

Suppose we want to keep the state name, 2019 property-crime rate, and 2019 burglary rate together for three states.

````python
states = ["Maryland", "Virginia", "Pennsylvania"]
property_rates = [1950.2, 1642.7, 1403.4]  # reported offenses per 100,000 people
burglary_rates = [278.9, 162.8, 182.4]    # reported offenses per 100,000 people

for i in range(len(states)):
    if states[i] == "Virginia":
        virginia_rate = property_rates[i]

virginia_rate
````

These lists depend on position. If one list is sorted, shortened, or extended without making the same change to the others, the values no longer describe the correct state.

````python
states_copy = states.copy()
property_rates_copy = property_rates.copy()

states_copy.sort()
states_copy
````

````python
property_rates_copy
````

The state names are now in alphabetical order, but the rates remain in their original order. The first positions no longer describe the same state.

## Part 3: DataFrames Keep Related Values Together

A pandas **DataFrame** is a table with named columns. Each row represents one observation, and the values in that row stay together when we sort or select data.

A DataFrame is similar to a spreadsheet, but we can inspect and analyze it with Python.

## Part 4: Load the Data

Pandas is a Python library for working with tabular data. We import it using the conventional abbreviation `pd`. Then `pd.read_csv()` reads a comma-separated values file and creates a DataFrame.

The classroom file below is a prepared subset of the larger State Crime dataset. Preparing it in advance lets us focus today on understanding a DataFrame. We will learn how to create subsets ourselves in the next class.

### What the Rows and Columns Mean

One row represents **one state in one year**. Because the file contains four states observed in five years, it has 20 state-year records.

| Column | Meaning |
|---|---|
| `state` | State represented by the row |
| `year` | Year represented by the row |
| `property_all` | Reported property-crime rate |
| `property_burglary` | Reported burglary rate |
| `property_larceny` | Reported larceny rate |
| `property_motor` | Reported motor-vehicle-theft rate |

All four crime measures are **rates per 100,000 people**, not raw crime counts. The denominator makes rates more useful than counts when state populations differ.

## Part 6: Check the Dataset's Coverage

The methods below answer different questions about a column:

- `.unique()` shows which values appear.
- `.nunique()` counts the number of different values.
- `.value_counts()` shows how often each value appears.
- `.value_counts(normalize=True)` shows the share of records with each value.

````python
df['state'].unique()
````

````python
df['state'].nunique()
````

````python
df['state'].value_counts()
````

````python
df['state'].value_counts(normalize=True)
````

Adding `normalize=True` divides each category count by the total number of records. Each state has 5 of the 20 records, so each state has a share of `0.25`, or 25%.

This is the share of records in this DataFrame. It is not a crime rate or a percentage of the state's population.

````python
df['year'].unique()
````

## Part 7: Select Columns

We select a column by placing its name in brackets. Selecting one column returns a pandas **Series**, which is one labeled column of values.

````python
property_rates = df['property_all']
property_rates
````

Pandas provides methods that calculate across a Series. Here, each result summarizes the 20 state-year rates in the column.

````python
# Maximum property-crime rate
property_rates.max()
````

````python
# Minimum property-crime rate
property_rates.min()
````

````python
# Mean of the 20 state-year rates
property_rates.mean()
````

To select multiple columns, put a list of column names inside the brackets. The result is another DataFrame.

````python
selected_columns = df[['state', 'year', 'property_all']]
selected_columns.head(n=8)
````

## Part 8: Sort Records

Sorting lets us put records in a useful order while keeping every row intact. By default, `.sort_values()` sorts from lowest to highest.

````python
lowest_first = df.sort_values(by='property_all')
lowest_first[['state', 'year', 'property_all']].head(n=5)
````

````python
highest_first = df.sort_values(by='property_all', ascending=False)
highest_first[['state', 'year', 'property_all']].head(n=5)
````

````python
# Sort first by state and then by year within each state
by_state_and_year = df.sort_values(by=['state', 'year'])
by_state_and_year[['state', 'year', 'property_all']].head(n=10)
````

## Part 9: Answer the Question

Our question was:

> Among four nearby states from 2015 through 2019, which state-year had the highest reported property-crime rate, which had the lowest, and how large was the difference?

````python
comparison = df[['state', 'year', 'property_all']].sort_values(by='property_all')
comparison.head(n=1)
````

````python
comparison.tail(n=1)
````

````python
difference = df['property_all'].max() - df['property_all'].min()
difference
````

### What the Results Show

Among the 20 state-year records, Pennsylvania in 2019 had the lowest reported property-crime rate at 1,403.4 per 100,000 people. Delaware in 2016 had the highest rate at 2,766.0 per 100,000. The difference between those two records was 1,362.6 reported property crimes per 100,000 people.

Notice that the answer identifies the records being compared, names the measure, and includes its unit. It does not describe the difference as a crime count or as a national rate.
