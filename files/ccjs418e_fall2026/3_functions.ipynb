## Part 1: Repeating the Same Calculation

The two lists below contain response times from two practice areas. Without a function, calculating both averages requires writing the same steps twice.

````python
area_a_response_times = [5, 12, 7, 9, 15, 8]
area_b_response_times = [6, 10, 8, 11, 13, 7]

area_a_total = 0
for response_time in area_a_response_times:
    area_a_total = area_a_total + response_time
area_a_average = area_a_total / len(area_a_response_times)

area_b_total = 0
for response_time in area_b_response_times:
    area_b_total = area_b_total + response_time
area_b_average = area_b_total / len(area_b_response_times)

print(f"Area A average: {area_a_average} minutes")
print(f"Area B average: {area_b_average} minutes")
````

The two calculations follow the same instructions:

1. Start a total at zero.
2. Add every response time to the total.
3. Divide by the number of response times.

A function lets us name those instructions once and use them with different lists.

## Part 2: Define and Call a Function

The function below accepts a list of response times, calculates the average, and returns the result.

````python
def average_response_time(response_times):
    total = 0
    for response_time in response_times:
        total = total + response_time
    average = total / len(response_times)
    return average
````

````python
area_a_average = average_response_time(area_a_response_times)
area_b_average = average_response_time(area_b_response_times)
difference = area_a_average - area_b_average

print(f"Area A average: {area_a_average} minutes")
print(f"Area B average: {area_b_average} minutes")
print(f"Difference: {difference} minutes")
````

### Anatomy of the Function

1. `def` begins the function definition.
2. `average_response_time` is the function's name.
3. `response_times` is a parameter: the input name used inside the function.
4. The indented lines are the function body.
5. `return` sends the calculated value back to the line that called the function.

Defining a function does not run it. The function runs when we call it with parentheses.

## Part 3: Parameters and Arguments

A **parameter** is an input name in the function definition. An **argument** is the actual value supplied when the function is called.

The next function has two parameters. Changing either argument can change the result.

````python
def response_status(response_minutes, threshold):
    if response_minutes > threshold:
        return "Over threshold"
    else:
        return "At or below threshold"

status_with_10_minute_threshold = response_status(12, 10)
status_with_15_minute_threshold = response_status(12, 15)

print(status_with_10_minute_threshold)
print(status_with_15_minute_threshold)
````

In `response_status(12, 10)`, `12` is the argument assigned to `response_minutes` and `10` is assigned to `threshold`. These are **positional arguments**: Python matches the first argument to the first parameter, the second argument to the second parameter, and so on. Their order matters.

### Keyword Arguments: Name Each Input

A **keyword argument** names the parameter when the function is called. This makes the role of each value visible in the call itself. Because the names establish the match, keyword arguments do not have to follow the parameter order.

````python
status_with_keywords = response_status(response_minutes=12, threshold=10)
status_with_reversed_keywords = response_status(threshold=10, response_minutes=12)

print(status_with_keywords)
print(status_with_reversed_keywords)
````

Both calls produce the same result. In the second call, the order changes, but `threshold=10` and `response_minutes=12` still tell Python exactly where each value belongs. Keyword arguments are especially useful when a function has several inputs or when two inputs could easily be confused.

## Part 4: `return` and `print` Are Different

`print()` displays something. `return` sends a value out of the function so that the rest of the program can store it, compare it, or use it in another calculation.

````python
def print_response_status(response_minutes, threshold):
    if response_minutes > threshold:
        print("Over threshold")
    else:
        print("At or below threshold")

printed_result = print_response_status(12, 10)
returned_result = response_status(12, 10)

print(f"Value from the print-only function: {printed_result}")
print(f"Value returned by response_status: {returned_result}")
````

The print-only function displays a message but gives back `None`. The function with `return` gives back a value that we can store in `returned_result` and use later.

## Part 5: A Function That Counts Matches

A function can combine parameters, a loop, a condition, a counter, and `return`.

````python
def count_matches(values, target):
    count = 0
    for value in values:
        if value == target:
            count = count + 1
    return count
````

````python
reported_incidents = [
    "theft", "assault", "theft", "burglary",
    "vandalism", "theft", "assault", "robbery"
]

theft_count = count_matches(reported_incidents, "theft")
assault_count = count_matches(reported_incidents, "assault")

print(f"Thefts: {theft_count}")
print(f"Assaults: {assault_count}")
````

## Part 6: Use Familiar Functions Inside a New Function

The body of a function can use functions you already know. `average_response_time()` already uses `len()`. The next function uses `min()` and `max()` to calculate the difference between the fastest and slowest response times.

````python
def response_time_range(response_times):
    fastest = min(response_times)
    slowest = max(response_times)
    return slowest - fastest

sample_response_times = [4, 7, 12, 5, 8, 15, 3, 11, 9]
time_range = response_time_range(response_times=sample_response_times)
print(f"Range: {time_range} minutes")
````

## Part 7: Return a List

A function can build a new list and return it. This function collects every response time above a threshold.

````python
def find_slow_responses(response_times, threshold):
    slow_responses = []
    for response_time in response_times:
        if response_time > threshold:
            slow_responses.append(response_time)
    return slow_responses

slow_responses = find_slow_responses(
    response_times=sample_response_times,
    threshold=10
)
print(f"Slow responses: {slow_responses}")
````

### A Zero Denominator

A function can check for an input that would make its calculation impossible.

````python
def calculate_rate(numerator, denominator):
    if denominator == 0:
        return None
    return (numerator / denominator) * 100

rate = calculate_rate(denominator=50, numerator=20)
print(rate)
````

### Returning Several Values, Calling Another Function, and Docstrings

A function can return more than one value and can call another function. A docstring records what the function does.

````python
def summarize_response_times(values):
    """Return the minimum, maximum, and average response time."""
    fastest = min(values)
    slowest = max(values)
    average = average_response_time(values)
    return fastest, slowest, average

fastest, slowest, average = summarize_response_times(area_a_response_times)

print(f"Fastest: {fastest}")
print(f"Slowest: {slowest}")
print(f"Average: {average}")
````

Names created inside a function are local to that function. In the previous cell, the `average` created inside `summarize_response_times()` is separate from the `average` variable that receives the returned value outside the function.
